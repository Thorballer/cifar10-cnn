```python
import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow import keras
from keras import regularizers, layers
from keras.optimizers import Adam
```

```python
def make_dataset(path, split, subset):
    return tf.keras.preprocessing.image_dataset_from_directory(
        path,
        validation_split=split,
        subset=subset,
        batch_size=64,
        image_size=(32, 32),
        seed=1,
        label_mode='categorical')
```

```python
train_ds = make_dataset('/kaggle/input/cifar10/cifar10/train', .2, 'training')
```

```python
val_ds = make_dataset('/kaggle/input/cifar10/cifar10/train', .2, 'validation')
```

```python
test_ds = make_dataset('/kaggle/input/cifar10/cifar10/test', None, None)
```

```python
model = keras.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(64, 3),
    layers.MaxPooling2D(),
    layers.BatchNormalization(),
    layers.Conv2D(128, 3),
    layers.MaxPooling2D(),
    layers.BatchNormalization(),
    layers.Conv2D(64, 3),
    layers.MaxPooling2D(),
    layers.BatchNormalization(),
    layers.Flatten(),
    layers.Dense(10, activation='softmax')
])
```

```python
model.summary()
```

```python
model.compile(
    optimizer=Adam(), loss='categorical_crossentropy', metrics=['accuracy']
)
```

```python
model.fit(train_ds, validation_data=val_ds, verbose=1, epochs=10)
```

```python
model.evaluate(test_ds, verbose=1)
```
